```python
import numpy as np
import pandas as pd

data = [
    ["Ram", 85, 90, 88],
    ["Babu", 78, 82, 80],
    ["Tharun", 92, 88, 95]
]

df = pd.DataFrame(data, columns=["Name", "Maths", "Science", "English"])
print(df)

#name   
Ram = np.mean(df.loc[df["Name"] == "Ram", ["Maths", "Science", "English"]])
print(f"The average of Ram is: {Ram}")
Babu = np.mean(df.loc[df["Name"] == "Babu", ["Maths", "Science", "English"]])
print(f"The average of Babu is: {Babu}")
Tharun = np.mean(df.loc[df["Name"] == "Tharun", ["Maths", "Science", "English"]])
print(f"The average of Tharun is: {Tharun}")

# Subject
Maths=np.mean(df["Maths"])
print(f"The average of Maths is: {Maths}")
Science=np.mean(df["Science"])
print(f"The average of Science is: {Science}")
English=np.mean(df["English"])
print(f"The average of English is: {English}")

```

```python
import matplotlib.pyplot as plt

month = ['Jan', 'Feb', 'Mar', 'Apr', 'May']
sales = [200, 300, 400, 500, 600]

# Line chart
plt.figure(figsize=(10, 5))
plt.plot(month, sales,marker='o',color='b')
plt.title('Monthly Sales')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.grid(True)
plt.show()

# Bar chart
plt.figure(figsize=(10, 5))
plt.bar(month, sales, color='orange', label='Sales')
plt.title('Monthly Sales')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.legend()
plt.grid(True)
plt.show()

# Scatter plot
plt.figure(figsize=(10, 5))
plt.scatter(month, sales, color='g',marker='x')
plt.title('Monthly Sales')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.grid(True)
plt.show()
```
